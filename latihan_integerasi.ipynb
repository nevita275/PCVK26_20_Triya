# Langkah (Step) 2:

Baca Image dari URLs - Read Image From URLs

--Bahasa--

Pada Langkah ini kita akan membaca image dari URLs, menampilkannya menggunakan OpenCV, perhatikan ada perbedaan ketika membuka Image dalam RGB dan BGR. Channel Warna default untuk OpenCV adalah BGR

--English--

In this step we will read Images from urls, and display them using openCV, please note the difference when reading image in RGB and BGR format. The default input color channels are in BGR format for openCV, RGB?

```python
# Membuat list untuk menyimpan url dari beberapa image
urls = ["https://iiif.lib.ncsu.edu/iiif/0052574/full/800,/0/default.jpg",
        "https://iiif.lib.ncsu.edu/iiif/0016007/full/800,/0/default.jpg",
        "https://unair.ac.id/wp-content/uploads/2020/12/Kucing.jpg"]

# baca dan tampilkan image
# loop pada tiap url image, beberapa image dapat disimpan pada list
for url in urls:
  image = io.imread(url)                           #read image
  image = cv.resize(image, (0,0), fx=0.5, fy=0.5)  #resize image to half size
  image_2 = cv.cvtColor(image, cv.COLOR_BGR2RGB)   #convert color to RGB
  final_frame = cv.hconcat((image, image_2))       #concatenate image
  cv2_imshow(final_frame)                          #show image
  print('\n')
```

# Langkah (Step) 3
melihat ukuran File image

```python
tinggi = image_2.shape[0]
lebar = image_2.shape[1]
print("resolusi image: tinggi x lebar = ",tinggi," x ", lebar)
cv2_imshow(image_2)
```

# Langkah (Step) 4
mengakses pixel dengan memberi garis diagonal menyilang

```python
image_2 = cv.cvtColor(image, cv.COLOR_BGR2RGB)
image_3 = cv.cvtColor(image, cv.COLOR_BGR2RGB)

# membuat garis horizontal ditengah image
for y in range (lebar):
  image_3[int((tinggi)/2),y] = [255,255,255]

final_frame = cv.hconcat((image_2, image_3))
cv2_imshow(final_frame)
```

# TUGAS PRAKTIKUM D2

1. Kode untuk mengambil gambar kucing dan menyesuaikan ukurannya


```python
# Mengambil gambar kucing untuk dikerjakan pada Tugas praktikum D2
url_tugas = "https://unair.ac.id/wp-content/uploads/2020/12/Kucing.jpg"

image_asli = io.imread(url_tugas)
image_asli = cv.resize(image_asli, (0,0), fx=0.5, fy=0.5)
img_rgb = cv.cvtColor(image_asli, cv.COLOR_BGR2RGB)

tinggi = img_rgb.shape[0]
lebar = img_rgb.shape[1]
```

2. Kode untuk membuat garis putih vertikal dan diagonal pada gambar kucing tersebut

```python

# 1. GARIS VERTIKAL
img_vertikal = img_rgb.copy() # Siapkan kertas kosong 1

# Membuat garis vertikal di tengah gambar
for y in range(tinggi):
    img_vertikal[y, int(lebar/2)] = [255, 255, 255]

# Menggabungkan gambar asli dengan gambar hasil vertikal
hasil_vertikal = cv.hconcat((img_rgb, img_vertikal))
print("1. Hasil Garis Vertikal:")
cv2_imshow(hasil_vertikal)
print('\n')

# 2. GARIS DIAGONAL
img_diagonal = img_rgb.copy() # Siapkan kertas kosong 2

# Membuat garis diagonal menyilang dari ujung ke ujung
for y in range(tinggi):
    titik_x1 = int(y * (lebar / tinggi))  # Garis dari kiri atas ke kanan bawah
    titik_x2 = lebar - 1 - titik_x1       # Garis dari kanan atas ke kiri bawah

    # Menggambar titik putihnya
    if titik_x1 < lebar:
        img_diagonal[y, titik_x1] = [255, 255, 255]
    if titik_x2 >= 0 and titik_x2 < lebar:
        img_diagonal[y, titik_x2] = [255, 255, 255]

# Menggabungkan gambar asli dengan gambar hasil diagonal
hasil_diagonal = cv.hconcat((img_rgb, img_diagonal))
print("2. Hasil Garis Menyilang Diagonal:")
cv2_imshow(hasil_diagonal)
```

3. Kode untuk membuat garis horisontal berwarna putih dibagian tengah gambar dengan panjang tertentu

```python
# 3. GARIS HORIZONTAL
# Menggunakan link gambar kucing berbeda
url_kucing_hitam = "https://images.alodokter.com/dk0z4ums3/image/upload/v1775008190/attached_image/5-fakta-kucing-hitam-ini-yang-perlu-diketahui-0-alodokter.jpg"
image_asli3 = io.imread(url_kucing_hitam)
image_asli3 = cv.resize(image_asli3, (0,0), fx=0.5, fy=0.5)
img_rgb3 = cv.cvtColor(image_asli3, cv.COLOR_BGR2RGB)

tinggi3 = img_rgb3.shape[0]
lebar3 = img_rgb3.shape[1]

# Proses pembuatan garis horizontal
img_horizontal = img_rgb3.copy()

titik_awal = int(lebar3 * 0.25)
titik_akhir = int(lebar3 * 0.75)

for y in range(titik_awal, titik_akhir):
  img_horizontal[int(tinggi3/2), y] = [255, 255, 255]

# Gabungkan dan Tampilkan
hasil_horizontal = cv.hconcat((img_rgb3, img_horizontal))
print("3. Hasil Garis Horizontal Sebagian:")
cv2_imshow(hasil_horizontal)
```

4. Kode untuk membuat kotak menggunakan kumpulan pixel warna putih di sembarang tempat dalam gambar

```python
# 4. KOTAK PUTIH BEBAS
img_kotak = img_rgb3.copy() #gambar dari no.3

# Mengubah warna piksel menjadi putih berbentuk kotak
# Format: [batas_tinggi, batas_lebar]
img_kotak[70:150, 110:200] = [255, 255, 255]

# Menggabungkan gambar asli dengan hasil kotak putih
hasil_kotak = cv.hconcat((img_rgb3, img_kotak))
print("4. Hasil Kotak Putih:")
cv2_imshow(hasil_kotak)
```
