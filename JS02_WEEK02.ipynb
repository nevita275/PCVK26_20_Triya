Label harus terlihat seimbang pada semua foto tanpa disetel satu per satu. Fungsi yang sama dikenakan pada dua citra berbeda ukuran. Tinggi bar dan ukuran huruf ikut menyesuaikan, sehingga proporsinya tetap sama. Silakan melakukan akuisisi data sendiri, setiap anak harus menggunakan image yang berbeda-beda**

```python
import cv2 as cv
import matplotlib.pyplot as plt

# 1. Buat sebagai Fungsi agar bisa dipakai berulang kali
def tambah_label_adaptif(image_path):
    img_doc = cv.imread(image_path)

    # Cek apakah file ada
    if img_doc is None:
        print(f"Peringatan: File '{image_path}' tidak ditemukan!")
        return None, 0, 0

    img_doc_rgb = cv.cvtColor(img_doc, cv.COLOR_BGR2RGB)
    h, w, c = img_doc_rgb.shape

    # 2. Rumus ADAPTIF
    tinggi_bar = h // 12
    skala_font = h / 900
    ketebalan = max(1, h // 400)

    # 3. Gambar Bar
    cv.rectangle(img_doc_rgb, (0, h - tinggi_bar), (w, h), color=(0, 0, 0), thickness=-1)

    # 4. Tambahkan Teks
    teks = "Dokumentasi PCVK - 2026"
    pos_x = w // 40
    pos_y = h - (tinggi_bar // 3)
    cv.putText(img_doc_rgb, teks, (pos_x, pos_y), cv.FONT_HERSHEY_SIMPLEX, skala_font, (255, 255, 255), ketebalan, cv.LINE_AA)

    return img_doc_rgb, w, h


file_landscape = 'kucing.png'
file_portrait = 'kucing2.jpeg'

# 5. Panggil fungsi yang sama untuk dua gambar berbeda
hasil_1, w1, h1 = tambah_label_adaptif(file_landscape)
hasil_2, w2, h2 = tambah_label_adaptif(file_portrait)

# 6. Tampilkan keduanya berdampingan
plt.figure(figsize=(12, 6))

if hasil_1 is not None:
    plt.subplot(1, 2, 1)
    plt.imshow(hasil_1)
    plt.title(f"Landscape ({w1} x {h1})")
    plt.axis('off')

if hasil_2 is not None:
    plt.subplot(1, 2, 2)
    plt.imshow(hasil_2)
    plt.title(f"Portrait ({w2} x {h2})")
    plt.axis('off')

plt.tight_layout()
plt.show()
```

Pada Studi kasus 1, program dirancang menggunakan sebuah fungsi khusus  (tambah_label_adaptif) agar tidak menggunakan angka mutlak yang kaku (hardcode) untuk menentukan ukuran elemen. Agar program bersifat adaptif terhadap semua jenis foto kamera dan ponsel, pengaturan tinggi bar hitam, posisi sumbu Y teks, ketebalan, dan ukuran font dihitung menggunakan operasi matematika pembagian  proporsional yang mengacu pada tinggi citra aslinya (h). Sebagai pembuktian, satu fungsi yang sama persis dipanggil untuk memproses dua citra yang bentuk dan resolusinya berbeda drastis (foto landscape kecil dan foto portrait besar). Hasilnya menunjukkan bahwa bar identitas di bagian bawah selalu menempel tepat di ujung bawah gambar dengan ketebalan dan ukuran teks yang secara otomatis menyesuaikan diri sehingga proporsinya tetap seimbang di kedua foto tersebut tanpa perlu disetel manual satu per satu.

**Studi Kasus 2: Marketplace seperti Tokopedia dan Shopee menampilkan foto produk dalam bingkai persegi 1:1. Foto penjual yang berukuran potrait atau lanskap akan terpotong atau terlihat tidak proporsional bila diunggah begitu saja. Bangun program yang mengubah foto berukuran apa pun menjadi thumbnail
persegi tanpa mengubah rasio aslinya, dengan sisa ruang diisi kanvas berwarna. Dalam hal ini, rasio asli tetap terjaga, citra terpasang tepat di tengah kanvas, tambahkan logo nama toko berukuran proporsional pada gambar.**

```python
import cv2 as cv
import numpy as np
import matplotlib.pyplot as plt

def buat_thumbnail_persegi(image_path, nama_toko):
    img = cv.imread(image_path)
    if img is None:
        print(f"Peringatan: File '{image_path}' tidak ditemukan!")
        return None
    img_rgb = cv.cvtColor(img, cv.COLOR_BGR2RGB)
    h, w, c = img_rgb.shape
    # Cari sisi terpanjang untuk patokan ukuran kanvas persegi (1:1)
    # Jika gambar memanjang, patokannya lebarnya. Jika meninggi, patokannya tingginya.
    s = max(h, w)
    # Buat Kanvas Kosong (Persegi)
    # np.ones bernilai 1 semua, dikali 220 menjadi warna putih total
    kanvas = np.ones((s, s, 3), dtype=np.uint8) * 220
    # Hitung titik koordinat agar gambar asli pas di tengah
    y_offset = (s - h) // 2
    x_offset = (s - w) // 2
    # Tempelkan gambar asli ke tengah kanvas
    kanvas[y_offset:y_offset+h, x_offset:x_offset+w] = img_rgb
    # Tambahkan Nama Toko
    skala_font = s / 800
    ketebalan = max(1, s // 400)
    pos_x = s // 30
    pos_y = s // 15
    # Tulis teks ke atas KANVAS
    cv.putText(kanvas, nama_toko, (pos_x, pos_y), cv.FONT_HERSHEY_DUPLEX, skala_font, (255, 0, 255), ketebalan, cv.LINE_AA)
    return kanvas
file_1 = 'kucing.png'     # Gambar bentuk mendatar (landscape)
file_2 = 'kucing2.jpeg'       # Gambar bentuk meninggi (portrait)
# Panggil fungsi untuk kedua gambar
thumb_1 = buat_thumbnail_persegi(file_1, "NevitaShop")
thumb_2 = buat_thumbnail_persegi(file_2, "NevitaShop")
# Tampilkan Hasilnya Berdampingan
plt.figure(figsize=(12, 6))
if thumb_1 is not None:
    plt.subplot(1, 2, 1)
    plt.imshow(thumb_1)
    plt.title("Thumbnail (Asal Landscape)")
    plt.axis('off')
if thumb_2 is not None:
    plt.subplot(1, 2, 2)
    plt.imshow(thumb_2)
    plt.title("Thumbnail (Asal Portrait)")
    plt.axis('off')
plt.tight_layout()
plt.show()
```

Dari hasil eksekusi program terhadap dua citra berbeda, terbukti program berjalan secara cerdas dan adaptif. Pada citra dengan format landscape, program secara otomatis mengisi sisa ruang kosong di bagian atas dan bawah dengan kanvas abu-abu. Sebaliknya, pada citra dengan format portrait, program mengisi sisa ruang di sisi kiri dan kanan. Keduanya berhasil menjadi bingkai persegi 1:1 yang sempurna dengan posisi teks yang proporsional tanpa merusak bentuk objek asli pada foto.
